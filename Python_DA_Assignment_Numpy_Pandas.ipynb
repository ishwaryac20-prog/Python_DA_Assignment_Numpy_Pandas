## Data Analysis Using NumPy 🔢

NumPy is a powerful Python library for numerical computing. Its main feature is the `ndarray`, which supports efficient numerical operations, indexing, slicing, and mathematical calculations.

### 1. Create a 1D NumPy Array 🌡️

Create a 1D NumPy array named `temperatures_w1` for Week 1 with the given temperature values.

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

temperatures_w1
```

### 2. Inspection and Properties 🔍

Inspect the shape, data type, and number of elements in the array.

```python
print("Inspection and Properties")
print()
print("Shape          :", temperatures_w1.shape)
print("Data type      :", temperatures_w1.dtype)
print("Num of elements:", temperatures_w1.size)
```

### 3. Array Operations 🌡️

Convert the Week 1 temperatures from Celsius to Fahrenheit.

**Formula:** `Fahrenheit = (Celsius × 9/5) + 32`

Then find the maximum, minimum, and mean temperatures.

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperature in Fahrenheit:", np.round(temperatures_f, 2))

print()
print("Max / Min / Mean")
print()
print("Maximum :", np.max(temperatures_f))
print("Minimum :", np.min(temperatures_f))
print("Mean    :", np.mean(temperatures_f))
```

### 4. Array Slicing and Indexing ✂️

Extract:
- The first three days
- The weekend (last two days)
- The middle three days

```python
print("Slicing and Indexing (Temperatures)")
print()
print("Temp for the first three days  :", temperatures_f[0:3])
print("Temp for the last two days     :", temperatures_f[-2:])
print("Temp for the middle three days :", temperatures_f[2:5])
```

### 5. Create a 2D NumPy Array 🧮

Create a 2D array named `temperatures`, where each row represents a different week.

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

temperatures
```

### 6. Inspect and Slice the 2D Array 🔍

Inspect the shape, data type, and total number of elements. Then extract the temperatures for each week and the weekend temperatures for both weeks.

```python
print("Inspect and Slice")
print()
print("Temp Shape             :", temperatures.shape)
print("Temp Data type         :", temperatures.dtype)
print("Temp (Num of elements) :", temperatures.size)

week_1 = temperatures[0]
week_2 = temperatures[1]

print()
print("====== Temperature for each week ======")
print("Week 1 :", week_1)
print("Week 2 :", week_2)

print()
print("====== Weekend (Last two days) ======")
print("Weekend 1 :", week_1[-2:])
print("Weekend 2 :", week_2[-2:])
```

## Pandas Series 🐼

A Pandas Series is a one-dimensional labeled array. It consists of values and index labels, making it useful for structured data analysis.

### 1. Creating Pandas Series 📝

Create a Pandas Series named `marks` with marks 95, 92, 89, 85, 80 and custom indexes `Rank1` to `Rank5`.

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Pandas Series")
print()
print(marks)
```

### 2. Indexing and Slicing 🔎

Use integer indexing, `loc`, `iloc`, and a Boolean mask to access and filter the Series.

```python
# Integer index position: 1st rank student
print("1st rank mark:", marks.iloc[0])

# loc: top 3 ranks using index labels
print()
print("Top 3 ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# iloc: 3rd-ranked student
print()
print("3rd rank mark:", marks.iloc[2])

# Boolean mask: marks greater than 90
print()
print("Ranks with marks greater than 90:")
print(marks[marks > 90])
```

### 3. Manipulating Series ✏️

Modify the first-rank mark, remove the last-rank entry, and compute CGPA by dividing each mark by 10.

```python
# Modify the mark of Rank1 to 100
marks['Rank1'] = 100

# Remove the last rank
marks = marks.drop('Rank5')

# Compute CGPA
CGPA = marks / 10

print("Marks:")
print(marks)

print()
print("CGPA:")
print(CGPA)
```

## Pandas DataFrame 📊

A Pandas DataFrame is a two-dimensional labeled data structure with rows and columns, similar to an Excel spreadsheet or SQL table.

### 1. Creating Pandas DataFrame 🛒

Create a DataFrame named `transactions` using the TransactionID, ProductCategory, Region, and Amount values provided in the assignment.

```python
transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": [
        "Electronics", "Clothing", "Electronics", "Furniture", "Clothing",
        "Electronics", "Furniture", "Clothing", "Furniture", "Electronics"
    ],
    "Region": [
        "North", "South", "North", "East", "West",
        "North", "East", "West", "South", "North"
    ],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

transactions
```

### 2. Data Exploration 🔍

Perform the following:
- Display the DataFrame using `head()` and `tail()`
- Check shape, column names, and data types
- Display `ProductCategory` and `Amount`
- Retrieve the last three columns
- Filter North-region transactions with Amount greater than 200
- Find product category value counts
- Find unique regions
- Find mean Amount by Region

```python
print("First 5 rows:")
display(transactions.head())

print("Last 5 rows:")
display(transactions.tail())

print("Shape:", transactions.shape)
print("Column names:", list(transactions.columns))

print()
print("Data types:")
print(transactions.dtypes)
```

```python
# Display ProductCategory and Amount
print("ProductCategory and Amount:")
display(transactions[['ProductCategory', 'Amount']])

# Retrieve the last 3 columns
print("Last 3 columns:")
display(transactions.iloc[:, -3:])

# Filter North region and Amount > 200
filter_df = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print("North region with Amount greater than 200:")
display(filter_df)

# Value counts
print("Product Category Value Counts:")
print(transactions['ProductCategory'].value_counts())

# Unique regions
print()
print("Unique Regions:")
print(transactions['Region'].unique())

# Mean amount by region
print()
print("Mean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

### 3. Manipulating the DataFrame ✏️

Modify the Amount for TransactionID 102, add a Discount column, remove TransactionID 109, and finally delete the Discount column.

```python
# 1. Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("After modifying TransactionID 102:")
display(transactions)

# 2. Add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

print("After adding Discount:")
display(transactions)

# 3. Remove TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

print("After removing TransactionID 109:")
display(transactions)

# 4. Delete Discount column
transactions = transactions.drop(columns=['Discount'])

print("Final DataFrame:")
display(transactions)
```
